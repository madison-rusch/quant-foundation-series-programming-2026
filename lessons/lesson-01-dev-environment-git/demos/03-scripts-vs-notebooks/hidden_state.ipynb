# Demo 3b — Hidden State in a Notebook

**Instructor:** run the cells **in order** first. Then re-run cell 2 twice more and watch the answer change.

This is the point of the section: the notebook *looks* correct, but a fresh top-to-bottom run would not reproduce what's on screen.

```python
# Cell 1 — starting data
prices = [101.2, 100.8, 102.5, 103.1, 102.7]
window = 3
len(prices)
```

```python
# Cell 2 — "just adding today's close"
# Run this cell twice and the list silently grows twice.
prices.append(104.0)
prices
```

```python
# Cell 3 — the analysis
averages = [
    sum(prices[i : i + window]) / window
    for i in range(len(prices) - window + 1)
]
[round(a, 2) for a in averages]
```

## Discussion

- Nothing here is *wrong* Python. The bug is the **execution order**, which the file does not record.
- `Kernel -> Restart & Run All` is the honesty check. If the output changes, the notebook was lying.
